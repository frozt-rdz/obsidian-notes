```python
def serie_anual(n=24, tasa=0.02, nivel=14.0):
    """Serie simulada sin ruido; un valor por año desde 2000, en °C."""
    valores = []  # Lista vacía: iremos añadiendo valores.
    for i in range(n):
        valores.append(nivel + tasa * i)
    return valores
```
